```python
# COMMAND ---------- NOTEBOOK

dbutils.widgets.text("catalog", "dev", "Target catalog")
dbutils.widgets.text("ingestion_pipeline_id", "d2959bb3-678c-4e0b-acb3-1597cd93bd31", "taklaproducts_ingestion_pipeline ID")
dbutils.widgets.text("transformation_pipeline_id", "08195499-fa49-4e57-a3e1-af60a4309a15", "taklaproducts_transformation_pipeline ID")

catalog = dbutils.widgets.get("catalog")
ingestion_pipeline_id = dbutils.widgets.get("ingestion_pipeline_id")
transformation_pipeline_id = dbutils.widgets.get("transformation_pipeline_id")

if not ingestion_pipeline_id or not transformation_pipeline_id:
    raise ValueError(
        "Both pipeline IDs are required. Find them in each pipeline's Settings panel "
        "in the Lakeflow Pipelines Editor — this notebook won't guess them."
    )


spark.sql(f"""
CREATE OR REPLACE VIEW {catalog}.taklaproducts.ingestion_event_log_raw AS
SELECT * FROM event_log("{ingestion_pipeline_id}")
""")

spark.sql(f"""
CREATE OR REPLACE VIEW {catalog}.taklaproducts.transformation_event_log_raw AS
SELECT * FROM event_log("{transformation_pipeline_id}")
""")

print(f"Created {catalog}.taklaproducts.ingestion_event_log_raw")
print(f"Created {catalog}.taklaproducts.transformation_event_log_raw")
```
